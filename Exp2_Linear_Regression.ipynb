**Step 1: Importing library**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step 2: Read the Data**

```python
data = {"hours":[1,2,3,4,5],
        "marks":[50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
x=df[["hours"]].values
y=df[["marks"]].values
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

```python
#"f-string" (formatted string literal) in Python
#.2f: This is a format specifier used within the curly braces of an f-string
print(f"Slope(m):{slope}")
print(f"Intercept(b):{intercept}")
print(f"Equatio of the regression line: y={slope[0]:.2f} x +{intercept[0]:.2f}")
```

```python
df['Predicted Marks']=model.predict(x)
```

**Visualizing**

```python
plt.scatter(df['hours'],df['marks'],color='blue',label='Actual Marks')
```

plotting the equation

```python
plt.plot(df['hours'],df['marks'],color='blue',label='Regression Line')
plt.xlabel('Hours studied')
plt.ylabel('Marks')
plt.title('Linear Regression: Hours vs Marks')
plt.legend()
plt.show()
```
